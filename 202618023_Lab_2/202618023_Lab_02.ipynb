## Task 2 - Vectorized Arithmetic and Linear Algebra

#### In this task, we use NumPy to perform vectorized arithmetic and basic linear algebra operations on matrices. We create two matrices and perform addition, element-wise multiplication, and matrix multiplication. We also find the transpose and determinant of a square matrix. If the matrix is invertible, we calculate its inverse and verify the result using `np.allclose()`. All operations are performed using NumPy's vectorized functions without using explicit Python loops.

```python
# Create two matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# Matrix Addition
C = A + B
print("\nMatrix C (A + B):")
print(C)

#Element-wise Multiplication

element_multiplication = A * B

print("Element-wise Multiplication:")
print(element_multiplication)

#Matrix Multiplication Use @ or np.matmul().

matrix_multiplication = A @ B

print("Matrix Multiplication:")
print(matrix_multiplication)

matrix_multiplication1= np.matmul(A, B)
print("Matrix Multiplication (using np.matmul):")
print(matrix_multiplication1)




```

```python
# Find the Transpose
A_transpose = A.T
print("Transpose of Matrix A:")
print(A_transpose)

#Find the Determinant
determinant_A = np.linalg.det(A)
print("Determinant of Matrix A:", determinant_A)    

```

```python
# Find the Inverse
inverse_A = np.linalg.inv(A)
print("Inverse of Matrix A:")
print(inverse_A)

#Verify A × A⁻¹ = I
identity_matrix = A @ inverse_A
print("A × A⁻¹:")
print(identity_matrix)

```

```python
#Use np.allclose() for Verification
identity = np.eye(2)
print("Is A @ A^-1 equal to Identity Matrix?")
print(np.allclose(A @ inverse_A, identity))
```

np.allclose() returns True because A @ A⁻¹ is approximately equal to the identity matrix. No explicit Python loops are used.

```python
#Use loc
print(df.loc[0, ["Name", "Age", "Sex"]])

print(df.loc[0:4, ["Name", "Age", "Sex"]])

#Use iloc
print(df.iloc[0, [1, 2, 3]])  # First row, columns 1, 2, 3

print(df.iloc[0:5, 1:4])  # First 5 rows, columns 1 to 3
```

Difference Between loc and iloc

loc is used to select rows and columns based on their labels or names. For example, df.loc[2, "Name"] selects the row with label 2 and the column named "Name".

iloc is used to select rows and columns based on their integer positions. For example, df.iloc[2, 1] selects the 3rd row and 2nd column.

In simple words:
loc uses labels, while iloc uses positions.

## Task 5 - Filtering and Querying

#### we use Boolean indexing and `query()` to filter the Titanic dataset based on different conditions. We calculate passenger counts and survival percentages for specific groups based on gender, age, class, fare, survival status, travel companions, and embarkation location.

```python
#1. Male passengers older than 50: How many?
male_passengers_older_than_50 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
count_male_passengers_older_than_50 = male_passengers_older_than_50.shape[0]
print("Number of male passengers older than 50:", count_male_passengers_older_than_50)

```

```python
# 2. Female first-class passengers
female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

number_female_first_class = len(female_first_class)
survival_percentage = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:", number_female_first_class)
print("   Percentage survived:", round(survival_percentage, 2), "%")
```

```python
# 3. Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()

result3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, Fare above median and survived:", len(result3))
print("   Overall median Fare:", fare_median)
```

```python
# 4. Travelling alone, age < 30, and did not survive
result4 = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30 and did not survive:", len(result4))
```

```python
# 5. Embarked = S, Pclass 2 or 3, Fare above Southampton median
southampton = df[df["Embarked"] == "S"]
southampton_median = southampton["Fare"].median()

result5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("\n5. Embarked S, Pclass 2/3 and Fare above Southampton median:", len(result5))
print("   Southampton median Fare:", southampton_median)
```

```python
result = df.query(
    "Age >= 20 and Age <= 40 and Fare > @fare_median and Survived == 1"
)

print(len(result))
```

## Task 6 - groupby() and Aggregation

#### we use `groupby()` and aggregation functions to analyze survival rates, passenger counts, average age, and fare based on Sex, Pclass, and Embarked groups.

```python
# 1. Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("1. Survival Rate by Sex:")
print(survival_by_sex.round(2))
```

```python
# 2. Survival rate by Pclass
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print("\n2. Survival Rate by Pclass:")
print(survival_by_class.round(2))
```

```python
# 3. Average Age and Fare by Pclass
average_by_class = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("\n3. Average Age and Fare by Pclass:")
print(average_by_class.round(2))
```

```python
# 4. Passenger count and survival rate by Sex-Pclass
sex_class = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

sex_class["Survival_Rate"] = sex_class["Survival_Rate"] * 100

print("\n4. Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_class.round(2))
```

```python
# 5. Passenger count, average Fare, and survival rate by Embarked
embarked_stats = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

embarked_stats["Survival_Rate"] = embarked_stats["Survival_Rate"] * 100

print("\n5. Passenger Count, Average Fare and Survival Rate by Embarked:")
print(embarked_stats.round(2))
```

## Task 7 - Missing Values and Fare Outliers

#### we identify and handle missing values using different imputation methods such as mean, median, mode, and random values. We also visualize missing values and detect Fare outliers using the IQR method.

```python
# 1. Missing-value count and percentage

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("Missing Value Information:")
print(missing_table.round(2))

# Plot missing values
plt.figure(figsize=(10, 5))

plt.bar(missing_count.index, missing_count.values)

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# 2. Fill missing Age with mean
print("\nMissing Age before filling:",
      df["Age"].isnull().sum())

mean_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(mean_age)

print("Mean Age:", mean_age)

print("Missing Age after filling:",
      df["Age"].isnull().sum())
```

```python
# 3. Different imputation methods

# Reload original dataset
df = pd.read_csv("train.csv")

# Mean
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(
    df_mean["Age"].mean()
)

print("\nMean Imputation:")
print("Missing Age:", df_mean["Age"].isnull().sum())


# Median
df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(
    df_median["Age"].median()
)

print("\nMedian Imputation:")
print("Missing Age:", df_median["Age"].isnull().sum())


# Mode
df_mode = df.copy()
df_mode["Age"] = df_mode["Age"].fillna(
    df_mode["Age"].mode()[0]
)

print("\nMode Imputation:")
print("Missing Age:", df_mode["Age"].isnull().sum())


# Random value
df_random = df.copy()

missing = df_random["Age"].isnull()

random_values = np.random.choice(
    df_random["Age"].dropna(),
    size=missing.sum()
)

df_random.loc[missing, "Age"] = random_values

print("\nRandom Imputation:")
print("Missing Age:", df_random["Age"].isnull().sum())
```

```python
# 4. Fare Outliers using IQR-

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

## Task 8 - Features and Pivot Table

#### we create new features such as `FamilySize` and `IsAlone` from existing columns. We also create a pivot table to compare the average survival rate by `Sex` and `Pclass` and identify the highest and lowest survival groups.

```python
#  Create FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print("FamilySize:")
print(df[["SibSp", "Parch", "FamilySize"]].head())


# Create IsAlone
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print("\nIsAlone:")
print(df[["FamilySize", "IsAlone"]].head())

#  Create pivot table
pivot = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("\nPivot Table - Mean Survival Rate:")
print(pivot.round(2))


# Convert survival rate to percentage
pivot_percentage = pivot * 100

print("\nPivot Table - Survival Percentage:")
print(pivot_percentage.round(2))
```

```python
# Find highest survival group
highest_group = pivot.stack().idxmax()
highest_rate = pivot.stack().max()

print("\nHighest Survival Group:")
print("Group:", highest_group)
print("Survival Rate:", round(highest_rate * 100, 2), "%")


# Find lowest survival group
lowest_group = pivot.stack().idxmin()
lowest_rate = pivot.stack().min()

print("\nLowest Survival Group:")
print("Group:", lowest_group)
print("Survival Rate:", round(lowest_rate * 100, 2), "%")
```
