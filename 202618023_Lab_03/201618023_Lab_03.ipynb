## Task 2 - Missing Values, Leakage, and Outliers



In this task, the dataset was checked for missing values and columns with very high missingness were identified. The `company` column was removed because most of its values were missing. Columns that could reveal the final booking outcome, such as `reservation_status` and `reservation_status_date`, were also removed to prevent data leakage. Numerical features were checked for extreme outliers using boxplots and the IQR method, and only clear/extreme outliers were removed.

### Missing-Value Count and Percentage

```python
# Calculate missing values
missing_count = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (missing_count / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Display missing values for every column
missing_summary
```

### Identify Columns with Very High Missingness

```python
# Display columns with missing values
missing_summary[missing_summary["Missing Percentage"] > 0].sort_values(
    "Missing Percentage", ascending=False
)
```

### Drop only the highly missing column

```python
# Drop company because it has very high missingness
X = X.drop("company", axis=1)

print("Updated X shape:", X.shape)
```

The company column contains missing values for the vast majority of bookings. Keeping a feature with approximately 94% missing data would provide very little useful information and could introduce unnecessary preprocessing complexity.

We do not drop agent, because only about 14% of its values are missing, so the remaining information can still be useful. It can be handled later using imputation.

### Remove Data Leakage Columns

```python
# Remove columns that directly reveal the final booking outcome
X = X.drop(
    ["reservation_status", "reservation_status_date"],
    axis=1
)

print("Updated X shape:", X.shape)
print("\nRemaining columns:")
print(X.columns.tolist())
```

### Check Numerical Features for Outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

print(numerical_cols)
```

```python
# Check outliers using boxplots

import matplotlib.pyplot as plt
import seaborn as sns
import math

# Get numerical columns
numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns

# Set number of columns in subplot
n_cols = 3
n_rows = math.ceil(len(numerical_cols) / n_cols)

# Create subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 4))

# Flatten axes
axes = axes.flatten()

# Create boxplots
for i, col in enumerate(numerical_cols):
    sns.boxplot(x=X[col], ax=axes[i])
    axes[i].set_title(f"Boxplot - {col}")

# Remove empty subplots
for i in range(len(numerical_cols), len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()
```

```python

# Calculate outliers using IQR
outlier_summary = []

for col in numerical_cols:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = X[(X[col] < lower_bound) | (X[col] > upper_bound)]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df
```

```python
outlier_df["Outlier Percentage"] = (
    outlier_df["Outlier Count"] / len(X) * 100
).round(2)

outlier_df
```

```python
# Make a copy before removing outliers
X_clean = X.copy()
y_clean = y.copy()

# Selected numerical features where extreme values can be meaningful
outlier_cols = [
    "lead_time",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies",
    "previous_cancellations",
    "booking_changes",
    "days_in_waiting_list",
    "adr",
    "required_car_parking_spaces",
    "total_of_special_requests"
]

# Create a mask for rows without extreme outliers
mask = pd.Series(True, index=X_clean.index)

for col in outlier_cols:
    Q1 = X_clean[col].quantile(0.25)
    Q3 = X_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR

    mask &= (X_clean[col] >= lower_bound) & (X_clean[col] <= upper_bound)

# Count rows removed
rows_removed = (~mask).sum()

# Remove the same rows from X and y
X_clean = X_clean[mask]
y_clean = y_clean[mask]

print("Rows removed:", rows_removed)
print("Original rows:", len(X))
print("Remaining rows:", len(X_clean))
```

## Task 3 - Create Two Preprocessing Pipelines



The data was split into training and testing sets using a stratified 80:20 split with `random_state=42`. Two preprocessing pipelines were created using `ColumnTransformer` and `Pipeline`. Numerical missing values were handled using KNN Imputation, while categorical missing values were replaced with the most frequent value and then one-hot encoded. Pipeline A uses `StandardScaler`, whereas Pipeline B uses `MinMaxScaler`. Preprocessing was fitted only on the training data to avoid data leakage and ensure a fair comparison.

### Train-Test Split

```python
pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_clean,
    y_clean,
    test_size=0.2,
    stratify=y_clean,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

### Numerical Missing-Value Imputation using KNN

```python
from sklearn.impute import KNNImputer

# Select numerical columns
numerical_cols = X_train.select_dtypes(include=["int64", "float64"]).columns

# Create KNN Imputer
knn_imputer = KNNImputer(n_neighbors=5)

# Fit on training data and transform
X_train[numerical_cols] = knn_imputer.fit_transform(
    X_train[numerical_cols]
)

# Transform test data using the same fitted imputer
X_test[numerical_cols] = knn_imputer.transform(
    X_test[numerical_cols]
)

print("Missing values in numerical columns after imputation:")
print(X_train[numerical_cols].isnull().sum())
```

```python
print("Train numerical nulls:", X_train[numerical_cols].isnull().sum().sum())
print("Test numerical nulls:", X_test[numerical_cols].isnull().sum().sum())
```

### Categorical Imputation and One-Hot Encoding

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_cols = X_train.select_dtypes(include=["object"]).columns

# Create imputer
cat_imputer = SimpleImputer(strategy="most_frequent")

# Impute missing categorical values
X_train[categorical_cols] = cat_imputer.fit_transform(
    X_train[categorical_cols]
)

X_test[categorical_cols] = cat_imputer.transform(
    X_test[categorical_cols]
)

# Create OneHotEncoder
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

# Fit on training data and transform both datasets
X_train_cat = encoder.fit_transform(X_train[categorical_cols])
X_test_cat = encoder.transform(X_test[categorical_cols])

print("Categorical data after encoding:")
print("X_train_cat shape:", X_train_cat.shape)
print("X_test_cat shape:", X_test_cat.shape)
```

### Pipeline A – KNNImputer + StandardScaler for Numerical Features

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

# Pipeline A: KNN Imputation + Standard Scaling
numeric_pipeline = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

# Fit on training data and transform both train and test
X_train_num = numeric_pipeline.fit_transform(X_train[numerical_cols])
X_test_num = numeric_pipeline.transform(X_test[numerical_cols])

print("X_train numerical shape:", X_train_num.shape)
print("X_test numerical shape:", X_test_num.shape)
```

### Pipeline B – KNNImputer + MinMaxScaler

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer
from sklearn.preprocessing import MinMaxScaler

# Pipeline B: KNN Imputation + Min-Max Scaling
numeric_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

# Fit only on training data
X_train_num_B = numeric_pipeline_B.fit_transform(
    X_train[numerical_cols]
)

# Transform test data using the same fitted pipeline
X_test_num_B = numeric_pipeline_B.transform(
    X_test[numerical_cols]
)

print("X_train numerical shape:", X_train_num_B.shape)
print("X_test numerical shape:", X_test_num_B.shape)
```

### ColumnTransformer + Pipeline A

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identify columns
numerical_cols = X_train.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = X_train.select_dtypes(include=["object"]).columns.tolist()

# Numerical Pipeline - Pipeline A
numeric_pipeline_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
print(categorical_pipeline)

# ColumnTransformer
preprocessor_A = ColumnTransformer([
    ("num", numeric_pipeline_A, numerical_cols),
    ("cat", categorical_pipeline, categorical_cols)
])

# Fit ONLY on training data
X_train_A = preprocessor_A.fit_transform(X_train)

# Transform test data using the fitted preprocessor
X_test_A = preprocessor_A.transform(X_test)

print("X_train shape:", X_train_A.shape)
print("X_test shape:", X_test_A.shape)



```

### ColumnTransformer + Pipeline B

```python
from sklearn.preprocessing import MinMaxScaler

# Numerical Pipeline - Pipeline B
numeric_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

# ColumnTransformer for Pipeline B
preprocessor_B = ColumnTransformer([
    ("num", numeric_pipeline_B, numerical_cols),
    ("cat", categorical_pipeline, categorical_cols)
])

# Fit ONLY on training data
X_train_B = preprocessor_B.fit_transform(X_train)

# Transform test data
X_test_B = preprocessor_B.transform(X_test)

print("X_train shape:", X_train_B.shape)
print("X_test shape:", X_test_B.shape)
```

## Task 5 - Evaluate and Compare the Four Results



The four trained model-pipeline combinations were evaluated using training accuracy, testing accuracy, precision, recall, and F1-score. A final comparison table was created to identify the best-performing models. Confusion matrices were plotted for the best Logistic Regression and Decision Tree results to understand their correct and incorrect predictions. The difference between training and testing accuracy was also checked to identify possible overfitting.

### Evaluate and Compare the Four Results

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Store all four experiments
models = {
    "Logistic Regression + Pipeline A": logistic_A,
    "Logistic Regression + Pipeline B": logistic_B,
    "Decision Tree + Pipeline A": tree_A,
    "Decision Tree + Pipeline B": tree_B
}

results = []

for name, model in models.items():

    # Predictions on training data
    y_train_pred = model.predict(X_train)

    # Predictions on testing data
    y_test_pred = model.predict(X_test)

    # Calculate metrics
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred)
    recall = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)

    results.append({
        "Experiment": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

# Create results table
results_df = pd.DataFrame(results)

# Round values for easy reading
results_df.iloc[:, 1:] = results_df.iloc[:, 1:].round(4)

results_df
```

```python
final_comparison = results_df.copy()

# Convert metric values to percentages
metric_cols = [
    "Training Accuracy",
    "Testing Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
]

final_comparison[metric_cols] = (
    final_comparison[metric_cols] * 100
).round(2)

display(final_comparison)
```

### Confusion Matrix for the Best Models

```python
# Find best Logistic Regression
logistic_results = results_df[
    results_df["Experiment"].str.contains("Logistic Regression")
]

best_logistic_name = logistic_results.loc[
    logistic_results["F1-Score"].idxmax(), "Experiment"
]

# Find best Decision Tree
tree_results = results_df[
    results_df["Experiment"].str.contains("Decision Tree")
]

best_tree_name = tree_results.loc[
    tree_results["F1-Score"].idxmax(), "Experiment"
]

print("Best Logistic Regression:", best_logistic_name)
print("Best Decision Tree:", best_tree_name)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Get the actual best models
best_logistic_model = models[best_logistic_name]
best_tree_model = models[best_tree_name]

# Predictions
y_pred_logistic = best_logistic_model.predict(X_test)
y_pred_tree = best_tree_model.predict(X_test)

# Confusion matrices
cm_logistic = confusion_matrix(y_test, y_pred_logistic)
cm_tree = confusion_matrix(y_test, y_pred_tree)

# Plot Logistic Regression confusion matrix
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Canceled", "Canceled"]
).plot(ax=axes[0], cmap="Blues", colorbar=False)

axes[0].set_title(f"Best Logistic Regression\n{best_logistic_name}")

# Plot Decision Tree confusion matrix
ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Not Canceled", "Canceled"]
).plot(ax=axes[1], cmap="Greens", colorbar=False)

axes[1].set_title(f"Best Decision Tree\n{best_tree_name}")

plt.tight_layout()
plt.show()
```

```python
# Calculate train-test accuracy difference
results_df["Accuracy Difference"] = (
    results_df["Training Accuracy"] -
    results_df["Testing Accuracy"]
).round(4)

display(results_df)
```

```python
for _, row in results_df.iterrows():
    difference = row["Accuracy Difference"]

    if difference < 0.02:
        comment = "Low difference → little evidence of overfitting."
    elif difference < 0.05:
        comment = "Moderate difference → some possible overfitting."
    else:
        comment = "Large difference → possible overfitting."

    print(f"{row['Experiment']}: {comment}")
```
